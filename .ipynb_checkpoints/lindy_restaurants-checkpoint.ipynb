```python
import polars as pl
import time
import requests
import hvplot
```

```python
df = pl.read_csv('dc_rests.csv',infer_schema_length=3000, ignore_errors=True)
```

```python
df['LICENSE_START_DATE'][0]
```

```python
# Convert the three date columns to pl.Datetime if needed
df = df.with_columns([
    pl.col("LICENSE_START_DATE").str.slice(0,10).alias("LICENSE_START_DATE"),
    pl.col("LICENSE_END_DATE").str.slice(0,10).alias("LICENSE_END_DATE"),
    pl.col("LICENSE_ISSUE_DATE").str.slice(0,10).alias("LICENSE_ISSUE_DATE"),
])
```

```python
# Assume your Polars DataFrame is named `df`

# Convert the three date columns to pl.Datetime if needed
df = df.with_columns([
    pl.col("LICENSE_START_DATE").str.to_date("%Y/%m/%d").alias("LICENSE_START_DATE"),
    pl.col("LICENSE_END_DATE").str.to_date("%Y/%m/%d").alias("LICENSE_END_DATE"),
    pl.col("LICENSE_ISSUE_DATE").str.to_date("%Y/%m/%d").alias("LICENSE_ISSUE_DATE"),
])
```

```python
# Filter to only abandoned licenses
df_abandoned = df.filter(pl.col("LICENSESTATUS") == "Abandoned")

# Add a column for the latest of the three dates
df_abandoned = df_abandoned.with_columns(
    pl.max_horizontal([
        pl.col("LICENSE_START_DATE"),
        pl.col("LICENSE_END_DATE"),
        pl.col("LICENSE_ISSUE_DATE")
    ]).alias("latest_date")
)

# Check how many times LICENSE_ISSUE_DATE is the latest
count_issue_latest = df_abandoned.filter(
    pl.col("LICENSE_ISSUE_DATE") == pl.col("latest_date")
).height

total_abandoned = df_abandoned.height

print(f"Issue date is the latest in {count_issue_latest} of {total_abandoned} abandoned licenses.")
```

```python
df = df.with_columns((pl.col("LICENSE_ISSUE_DATE")  - pl.col("LICENSE_START_DATE") ).dt.total_days().alias('days_to_abandon'))
```

```python
df = df.with_columns((pl.col("LICENSE_END_DATE")  - pl.col("LICENSE_START_DATE") ).dt.total_days().alias('days_open'))
```

```python
df['days_to_abandon'].hvplot.hist()
```
