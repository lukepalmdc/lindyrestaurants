```python
import polars as pl

def safe_json_clean(s: str) -> str:
    # Handle empty, None, or malformed cases
    if not s or not isinstance(s, str):
        return '{}'
    try:
        # Replace single with double quotes and validate
        s_clean = s.replace("'", '"')
        # Try decoding to confirm it's valid
        import json
        json.loads(s_clean)
        return s_clean
    except Exception:
        return '{}'


# Clean strings first using Python apply
df = df.with_columns(
    pl.col("coordinates")
    .map_elements(safe_json_clean)
    .alias("json_clean")
)

# Now parse safely
df = df.with_columns(
    pl.col("json_clean").str.json_decode().alias("geo_dict")
)

# Extract fields
df = df.with_columns([
    pl.col("geo_dict").struct.field("latitude").alias("latitude"),
    pl.col("geo_dict").struct.field("longitude").alias("longitude")
])

print(df)
```
