```python
import requests

headers = {
    "accept": "application/json",
    "authorization": "Bearer 3nFsqYY_7VQWn-Cz3DJQj7R-pZbjIqkwM2mNO_FoDydqF7vVklwEckAESVQCIchqnN3RDj3CVCkusZ_qAUdUeIQOm_8HsWptJo3hFsCh_wd1tuCEMLs5S-uy7gjuZ3Yx"
}
BASE_URL = "https://api.yelp.com/v3/businesses/search"
```

```python
dfs = []
```

```python
zips = [
    20001, 20002, 20003, 20004, 20005, 20006, 20007, 20008, 20009, 20010,
    20011, 20012, 20015, 20016, 20017, 20018, 20019, 20020, 20024, 20032,
    20036, 20037, 20045, 20052, 20057, 20059, 20064, 20204, 20220, 20230,
    20240, 20245, 20502, 20510, 20515, 20520, 20530, 20535, 20540, 20542,
    20551, 20560, 20565, 20566, 20591
]
```

```python
def get_restaurants(city, category="restaurants", max_results=200):
    restaurants = []
    limit = 50  # Max allowed by Yelp

    for offset in range(0, min(max_results, 200), limit):
        params = {"location": city, "categories": category, "limit": limit, "offset": offset, 'sort_by':'best_match'}
        response = requests.get(BASE_URL, headers=headers, params=params)

        if response.status_code != 200:
            print(f"Error: {response.json()}")  # Print error details
            break

        data = response.json()
        businesses = data.get("businesses", [])

        if not businesses:
            break  # Stop if no more businesses are returned
        
        restaurants.extend(businesses)
        print(f"Fetched {len(businesses)} businesses, offset {offset}")  # Debugging info
        
        time.sleep(1)  # Avoid rate limit

    return pl.DataFrame(restaurants)

def zip_code_rests():
    for zip_code in zips:
        zip_rests = get_restaurants(f"{zip_code}")
        dfs.append(zip_rests) 
    
zip_code_rests()
print(f"Total restaurants fetched: {len(restaurants)}")
```

```python
all_columns = set()
for df in dfs:
    all_columns.update(df.columns)
    
# Ensure all DataFrames have the same columns
normalized_dfs = []
for df in dfs:
    if not df.is_empty():
        # Fill missing columns with None
        df = df.with_columns([pl.lit(None).alias(col) for col in all_columns if col not in df.columns])
        # Reorder columns to match the full set
        df = df.select(sorted(all_columns))  
        normalized_dfs.append(df)

# Vertically stack DataFrames
if normalized_dfs:
    result_df = pl.concat(normalized_dfs, how="vertical_relaxed")
```
